```python
#Working on DT
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import f1_score, classification_report, confusion_matrix
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split,GridSearchCV
```

```python
#Baseline (unpruned) tree
dt = DecisionTreeClassifier(random_state=42)
dt.fit(X_train, y_train)
print("Train F1:", f1_score(y_train, dt.predict(X_train)))
print("Test F1:", f1_score(y_test, dt.predict(X_test)))
```

```python
#prepruning
params = {
    'max_depth': [3, 4, 5, 6, 8, 10],
    'min_samples_split': [10, 20, 50],
    'min_samples_leaf': [5, 10, 20],
    'criterion': ['gini', 'entropy'],
    'class_weight': ['balanced']   # helps with imbalance
}
grid = GridSearchCV(DecisionTreeClassifier(random_state=42),
                    params, scoring='f1', cv=5, n_jobs=-1)
grid.fit(X_train, y_train)
best = grid.best_estimator_
print("Best F1:", grid.best_score_)
print("Best params:", grid.best_params_)
```

```python
#Postprunning
path = dt.cost_complexity_pruning_path(X_train, y_train)
alphas = path.ccp_alphas
scores = []
for a in alphas:
    m = DecisionTreeClassifier(random_state=42, ccp_alpha=a,
                               class_weight='balanced').fit(X_train, y_train)
    scores.append(f1_score(y_test, m.predict(X_test)))
```

```python
y_pred = best.predict(X_test)
print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
print("Final F1:", f1_score(y_test, y_pred))
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import cross_val_score
from sklearn.metrics import f1_score, precision_score, recall_score

# ------------------------------------------------------------
# 1. Unpruned (baseline) tree
# ------------------------------------------------------------
dt = DecisionTreeClassifier(random_state=42)
dt.fit(X_train, y_train)

# ------------------------------------------------------------
# 2. Pre-pruned tree = 'best' (tumhara GridSearchCV wala model)
# ------------------------------------------------------------

# ------------------------------------------------------------
# 3. Post-pruned tree (cost-complexity pruning)
#    ccp_alpha ko TRAIN set par cross-validation se choose karte hain
#    (test set use nahi hota, isliye leakage nahi hoti)
# ------------------------------------------------------------
path = dt.cost_complexity_pruning_path(X_train, y_train)
alphas = path.ccp_alphas[:-1]            # last alpha = sirf root node, ignore

# Alphas bahut ho sakte hain, isliye ~30 evenly spaced values lete hain (speed ke liye)
idx = np.linspace(0, len(alphas) - 1, 30).astype(int)
alphas = np.unique(alphas[idx])

cv_scores = []
for a in alphas:
    m = DecisionTreeClassifier(random_state=42, ccp_alpha=a,
                               class_weight='balanced')
    cv_scores.append(cross_val_score(m, X_train, y_train,
                                     cv=5, scoring='f1', n_jobs=-1).mean())

best_alpha = alphas[int(np.argmax(cv_scores))]
print("Best ccp_alpha:", best_alpha, "| CV F1:", max(cv_scores))

post = DecisionTreeClassifier(random_state=42, ccp_alpha=best_alpha,
                              class_weight='balanced')
post.fit(X_train, y_train)

# Alpha vs F1 plot
plt.figure(figsize=(7, 4))
plt.plot(alphas, cv_scores, marker='o')
plt.axvline(best_alpha, color='red', linestyle='--', label=f'best alpha = {best_alpha:.5f}')
plt.xlabel("ccp_alpha")
plt.ylabel("CV F1 score")
plt.title("Post-pruning: alpha vs F1")
plt.legend()
plt.show()

# ------------------------------------------------------------
# 4. Comparison table
# ------------------------------------------------------------
models = {
    "Unpruned": dt,
    "Pre-pruned (GridSearch)": best,
    "Post-pruned (ccp_alpha)": post,
}

rows = []
for name, m in models.items():
    tr_pred = m.predict(X_train)
    te_pred = m.predict(X_test)
    rows.append({
        "Model": name,
        "Train F1": round(f1_score(y_train, tr_pred), 3),
        "Test F1": round(f1_score(y_test, te_pred), 3),
        "Gap (Train-Test)": round(f1_score(y_train, tr_pred) - f1_score(y_test, te_pred), 3),
        "Test Precision": round(precision_score(y_test, te_pred), 3),
        "Test Recall": round(recall_score(y_test, te_pred), 3),
        "Depth": m.get_depth(),
        "Leaves": m.get_n_leaves(),
        "Beats 0.55?": "Yes" if f1_score(y_test, te_pred) > 0.55 else "No",
    })

comparison = pd.DataFrame(rows).set_index("Model")
print(comparison)

# ------------------------------------------------------------
# 5. Bar chart of Test F1
# ------------------------------------------------------------
ax = comparison["Test F1"].plot(kind="bar", figsize=(7, 4), rot=0, color=["#d9534f", "#5bc0de", "#5cb85c"])
ax.axhline(0.55, color="black", linestyle="--", label="Benchmark 0.55")
ax.set_ylabel("Test F1")
ax.set_title("Unpruned vs Pre-pruned vs Post-pruned")
ax.legend()
plt.show()
```
